```python
import pandas as pd


dosya_yolu = '/kaggle/input/datasets/cokunkaanaki/drag-and-drop/ecommerce_sales_analytics_5000.csv'


df = pd.read_csv(dosya_yolu)
```

```python
df.info()
print("/n--------------------------------/n")

print(df.isnull().sum())
```

```python
display(df.describe())

print("/n--------------sutun-isimleri--------/n")
print(df.columns)
```

```python
kategori_ciro = df.groupby('product_category')['revenue'].sum().sort_values(ascending=False)

print("/n--------------kategori_ciro--------/n")
print(kategori_ciro)
```

```python
import matplotlib.pyplot as plt 

kategori_ciro.plot(kind='bar',color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'], figsize=(8,5))
plt.title('kategori bazlı toplm cirt')
plt.ylabel('toplam ciro(USD)')
plt.xlabel('kategori') 

plt.xticks(rotation=45) 

plt.show()  
```

```python
teslimat_etkisi = df.groupby('delivery_days')['customer_rating'].mean()

print("/n--------------teslimat_etkisi--------/n")
print(teslimat_etkisi)  

teslimat_etkisi.plot(kind='line', marker='o', color='orange', figsize=(8,5),grid=True )
plt.title('teslimat süresinin müşteri puanına etkisi')
plt.xlabel('teslimat süresi (gün)')
plt.ylabel('ortalama müşteri puanı')    
plt.show()      
```

```python
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression

x=df[['delivery_days', 'discount', 'unit_price']]
y=df['customer_rating'] 

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(x_train, y_train)

print("Model eğitildi. Katsayılar:")
print(f"teslimat süresinin etkisi: {model.coef_[0]}")
print(f"indirim miktarının etkisi: {model.coef_[1]}")
print(f"birim fiyatın etkisi: {model.coef_[2]}")
```

```python
import pandas as pd
yeni_siparis = pd.DataFrame({
    'delivery_days': [8],
    'discount': [0.05],
    'unit_price': [1200]
})

tahmini_musteri_puani = model.predict(yeni_siparis)

print(f"beklenen müşteri puanı: {tahmini_musteri_puani[0]:.2f} / 5.00")
```
